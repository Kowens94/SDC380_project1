```python
from datetime import datetime
import pandas as pd
from IPython.display import display, Markdown
# Load the dataset into a new DataFrame
df_new = pd.read_excel('C:\\Users\\owens\\Downloads\\AirBnBSummary_v2.xlsx')

import pandas as pd #import pandas library
todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "kenowe0432"
display(Markdown(f"### {student_id} – {todays_date}")) #display Header Text

total_listings = df_new['id'].count()
min_price = df_new['price'].min()
max_price = df_new['price'].max()
mean_price = df_new['price'].mean()
median_reviews = df_new['number_of_reviews'].median()
mode_nights = df_new['minimum_nights'].mode()
std_price = df_new['price'].std()
correlation_price_availability = df_new['price'].corr(df_new['availability_365'])

#Display Calculated Statistics
print("Total Listings:", total_listings)
print("Minimum Price:", min_price)
print("Maximum Price:", max_price)
print("Mean Price:",mean_price)
print("Median Reviews:", median_reviews)
print("Mode Nights:", mode_nights)
print("Standard Deviation Of Price:", std_price)
print("Correlation Between Price And Availability:", correlation_price_availability)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from datetime import datetime
student_id = "kenowe0432"
# Load the dataset into a new DataFrame
df_new = pd.read_excel('C:\\Users\\owens\\Downloads\\AirBnBSummary_v2.xlsx')
# Get current date and time
current_datetime = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
plt.figure(figsize=(10, 6))
plt.hist(df_new['number_of_reviews'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Number of Reviews')
plt.ylabel('Listings')
plt.title(f'Histogram of Number of Reviews vs. Listings - StudentID: {student_id} - {current_datetime}')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 8))
plt.barh(df_new['minimum_nights'],df_new['id'], color='green')
plt.xlabel('ID')
plt.ylabel('Minimum Nights')
plt.title(f'Minimum Nights - StudentID: {student_id} - {current_datetime}')
plt.gca().invert_yaxis()
plt.grid(True)
plt.show()
```

```python
x = df_new['availability_365']
y = df_new['minimum_nights']


plt.figure(figsize=(12, 8))
plt.scatter(x, y, label='Data Points', color='blue')

plt.xlabel('Availability')
plt.ylabel('Minimum Nights')

plt.title(f'365 Availability - StudentID: {student_id} - {current_datetime}')
plt.legend()
plt.grid(True)
```

```python
selected_columns = ['price', 'minimum_nights', 'number_of_reviews','availability_365']
correlation_matrix = df_new[selected_columns].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title(f'Heatmap of Correlations - StudentID: {student_id} - {current_datetime}')
plt.show()
```
